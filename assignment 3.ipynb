```python
import random

# 1. Set Up the Game
secret_number = random.randint(1, 10)

print("--- Welcome to the Number Guessing Game! ---")

# While loop & Control Statements
while True:
    try:
        # 2. Prompt the User
        guess = int(input("Enter your guess (1-10): "))
        
        # Control statement: break
        if guess == 0:
            print("You exited the game.")
            break
            
        # Control statement: continue
        if guess < 1 or guess > 10:
            print("Invalid range! Please guess between 1 and 10.")
            continue
            
        if guess == secret_number:
            print(f"Congratulations! You guessed the correct number: {secret_number}")
            break
        elif guess < secret_number:
            print("Too low! Try a higher number.")
        else:
            print("Too high! Try a lower number.")
            
    except ValueError:
        print("Invalid input! Please enter a valid number.")
        continue
else:
    print("Thanks for playing!")
```

```python
attempts = 3

while attempts > 0:
    guess = int(input("Enter your guess (1-10): "))
    
    if guess < 1 or guess > 10:
        print("Out of valid range! Please guess between 1 and 10.")
    elif guess > secret_number:
        print("Too high! The secret number is lower.")
    elif guess < secret_number:
        print("Too low! The secret number is higher.")
    else:
        print(f"Congratulations! You guessed the correct number: {secret_number}")
        break
        
    attempts -= 1
    print(f"Attempts left: {attempts}")

if attempts == 0:
    print(f"Out of attempts! The correct number was {secret_number}.")
```
